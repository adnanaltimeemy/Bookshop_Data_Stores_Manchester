```python
import pandas as pd

# List of H. P. Lovecraft books to analyze
books = [
    "The Call of Cthulhu",
    "At the Mountains of Madness",
    "The Dunwich Horror",
    "The Shadow over Innsmouth",
    "Collected Weird Tales"
]
```

```python
data = {
    "Title": books,
    "Waterstones_WeeklySales": [15, 10, 8, 5, 20],  # simulated weekly units
    "Indie_WeeklySales":    [8, 6, 4, 3, 12],       # smaller indie sales
    "Avg_Price_GBP":        [10.99, 12.50, 8.99, 9.99, 15.00]
}

df = pd.DataFrame(data)
```

```python
# compute revenues per week and per year (52 weeks)
df["Waterstones_WeeklyRevenue"] = df["Waterstones_WeeklySales"] * df["Avg_Price_GBP"]
df["Indie_WeeklyRevenue"]      = df["Indie_WeeklySales"]    * df["Avg_Price_GBP"]

df["Waterstones_YearlyRevenue"] = df["Waterstones_WeeklyRevenue"] * 52
df["Indie_YearlyRevenue"]      = df["Indie_WeeklyRevenue"]    * 52

df
```

```python
# Example demographic context (Manchester metrics in narrative)
manchester_population = 553230  # approximate Greater Manchester
students_aged_18_25 = 0.15     # simulated percent of population
```

```python
# Example cost/revenue assumptions
waterstones_margin = 0.25   # chain net margin as guess
indie_margin      = 0.40

df["Waterstones_WeeklyProfit"] = df["Waterstones_WeeklyRevenue"] * waterstones_margin
df["Indie_WeeklyProfit"]      = df["Indie_WeeklyRevenue"] * indie_margin

df["Waterstones_YearlyProfit"] = df["Waterstones_YearlyRevenue"] * waterstones_margin
df["Indie_YearlyProfit"]      = df["Indie_YearlyRevenue"] * indie_margin

df
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Title",
        y=["Waterstones_WeeklyRevenue","Indie_WeeklyRevenue"])
plt.title("Weekly Revenue Comparison")
plt.ylabel("Revenue (GBP)")
plt.show()
```
