```python
import pandas as pd
```

```python
data = [
    {"store":"Waterstones Deansgate", "type":"Chain/Large", "location":"City Centre Flagship"},
    {"store":"Waterstones Arndale", "type":"Chain/Large", "location":"High Footfall Mall"},
    {"store":"Blackwell’s Manchester", "type":"Chain/Large", "location":"University District"},
    {"store":"House of Books & Friends", "type":"Independent", "location":"Community Bookshop Café"},
    {"store":"Chapter One Books", "type":"Independent", "location":"Café + Books Venue"},
    {"store":"Chorlton Bookshop", "type":"Independent", "location":"Local Community Bookshop"},
    {"store":"Queer Lit", "type":"Independent", "location":"Specialist/LGBTQ+ Books"},
    {"store":"Catalog Bookshop", "type":"Independent", "location":"Alternative Book Selection"},
    {"store":"Didsbury Village Bookshop", "type":"Independent", "location":"Village Bookshop/Café"},
    {"store":"E J Morten Booksellers", "type":"Independent", "location":"Specialist/Rare Books"}
]

df = pd.DataFrame(data)
df
```

```python
df["avg_daily_customers"] = [
    1100, 1400, 900,   # chain stores
    200, 180, 170,     # independents
    120, 130, 140, 90
]

df["avg_spend_gbp"] = [
    24, 20, 22,
    23, 21, 20,
    28, 18, 19, 30
]

df["weekly_rent_gbp"] = [
    34000, 33000, 25000,
    7000, 6500, 5800,
    5000, 5200, 5500, 4000
]

df["weekly_other_costs_gbp"] = [
    10000, 9000, 8000,
    2500, 2800, 2200,
    2000, 1800, 2100, 1500
]

df
```

```python
DAYS_PER_WEEK = 7

df["weekly_revenue_gbp"] = (
    df["avg_daily_customers"] * DAYS_PER_WEEK * df["avg_spend_gbp"]
)
```

```python
df["gross_margin_pct"] = df["type"].apply(lambda x: 0.45 if x=="Chain/Large" else 0.55)
df["weekly_gross_profit_gbp"] = df["weekly_revenue_gbp"] * df["gross_margin_pct"]
```

```python
df["weekly_costs_total_gbp"] = df["weekly_rent_gbp"] + df["weekly_other_costs_gbp"]

df["weekly_net_profit_gbp"] = (
    df["weekly_gross_profit_gbp"] - df["weekly_costs_total_gbp"]
)

WEEKS_PER_YEAR = 52
df["yearly_net_profit_gbp"] = df["weekly_net_profit_gbp"] * WEEKS_PER_YEAR

df[[
    "store",
    "type",
    "weekly_revenue_gbp",
    "weekly_net_profit_gbp",
    "yearly_net_profit_gbp"
]]
```

```python
df.groupby("type")[["weekly_revenue_gbp", "weekly_net_profit_gbp"]].mean()
```
