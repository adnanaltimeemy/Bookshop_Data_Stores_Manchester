```python
import pandas as pd

# Structure for Book Retailers in Manchester
columns = [
    "Retailer",        # WHSmith or Waterstones
    "Location",        # Store area (e.g. Manchester City Centre)
    "Book_Category",   # Fiction, Non-Fiction, Children’s, Academic
    "Weekly_Sales",    # Estimated weekly unit sales
    "Avg_Price",       # Average sales price
    "Weekly_Revenue",  # Sales * Avg_Price
    "Customer_Demo",   # dominant customer profile (e.g., Students, commuters)
    "Weekly_Profits",  # revenue – estimated cost
]

df = pd.DataFrame(columns=columns)
```

```python
# Example Hypothetical Data for WHSmith Manchester branch
df.loc[len(df)] = [
    "WHSmith", "Manchester City Centre", "General Books",
    350,  # Weekly estimated units
    12.50, # Avg price per book
    350 * 12.50,
    "Commuters + Students", 
    (350 * 12.50) * 0.25  # assume 25% margin
]

df.loc[len(df)] = [
    "WHSmith", "Manchester City Centre", "Children’s Books",
    150, 8.00,
    150 * 8.00,
    "Families", (150 * 8.00) * 0.25
]
```

```python
# Example Hypothetical Data for Waterstones Manchester
df.loc[len(df)] = [
    "Waterstones", "Manchester City Centre", "General Books",
    550,
    14.00,
    550 * 14.00,
    "Book Lovers + Adults",
    (550 * 14.00) * 0.35  # higher margin on specialist books
]

df.loc[len(df)] = [
    "Waterstones", "Manchester City Centre", "Academic Books",
    200, 22.00,
    200 * 22.00,
    "University Students",
    (200 * 22.00) * 0.35
]
```

```python
df["Weekly_Revenue"] = df["Weekly_Sales"] * df["Avg_Price"]
df["Annual_Revenue"] = df["Weekly_Revenue"] * 52
df["Annual_Profit"] = df["Weekly_Profits"] * 52

df
```

```python
import matplotlib.pyplot as plt

grouped = df.groupby("Retailer")[["Weekly_Revenue","Annual_Revenue","Annual_Profit"]].sum()
grouped.plot(kind="bar")
plt.title("Manchester Book Retailers Revenue & Profit")
plt.ylabel("£ GBP")
plt.show()
```
