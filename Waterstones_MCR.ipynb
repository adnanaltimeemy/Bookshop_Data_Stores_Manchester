```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
waterstones_stores = pd.DataFrame([
    {"Store": "Waterstones Deansgate", "Location": "Manchester City Centre"},
    {"Store": "Waterstones Arndale Centre", "Location": "Manchester City Centre"},
    {"Store": "Waterstones Trafford Centre", "Location": "Trafford Centre"}
])
```

```python
categories = ["Fiction", "Nonfiction", "Children", "Stationery", "Events"]

waterstones_menu = pd.DataFrame({
    "Category": categories,
    "Price_Average_GBP": [10, 15, 8, 5, np.nan],
    "Inventory_Count": [12000, 8000, 5000, 2000, 50]
})

independent_menu = pd.DataFrame({
    "Category": categories,
    "Price_Average_GBP": [12, 18, 9, 6, np.nan],
    "Inventory_Count": [8000, 4000, 2000, 1000, 25]
})
```

```python
waterstones_stores["Est_Weekly_Sales_GBP"] = 34000
waterstones_stores["Est_Yearly_Sales_GBP"] = waterstones_stores["Est_Weekly_Sales_GBP"] * 52
```

```python
demographics = pd.DataFrame([
    {"Age_Group": "18-24", "Percentage": 13},
    {"Age_Group": "25-34", "Percentage": 20},
    {"Age_Group": "35-54", "Percentage": 30},
    {"Age_Group": "55+", "Percentage": 37}
])
```

```python
profit_margin = 0.08  # 8% estimate
waterstones_stores["Est_Weekly_Profit_GBP"] = waterstones_stores["Est_Weekly_Sales_GBP"] * profit_margin
waterstones_stores["Est_Yearly_Profit_GBP"] = waterstones_stores["Est_Yearly_Sales_GBP"] * profit_margin
```

```python
independent_stores = pd.DataFrame([
    {"Store": "House of Books & Friends", "Est_Weekly_Sales_GBP": 15000},
    {"Store": "Queer Lit", "Est_Weekly_Sales_GBP": 12000},
    {"Store": "Chorlton Bookshop", "Est_Weekly_Sales_GBP": 10000},
    {"Store": "Catalog Bookshop", "Est_Weekly_Sales_GBP": 8000}
])

independent_stores["Est_Yearly_Sales_GBP"] = independent_stores["Est_Weekly_Sales_GBP"] * 52
independent_stores["Est_Yearly_Profit_GBP"] = independent_stores["Est_Yearly_Sales_GBP"] * 0.1  # slightly higher margin
```

```python
def total_sales(df):
    return df["Est_Yearly_Sales_GBP"].sum()

def total_profit(df):
    return df["Est_Yearly_Profit_GBP"].sum()

print("Waterstones Yearly Sales", total_sales(waterstones_stores))
print("Waterstones Yearly Profit", total_profit(waterstones_stores))
print("Independents Yearly Sales", total_sales(independent_stores))
print("Independents Yearly Profit", total_profit(independent_stores))
```
